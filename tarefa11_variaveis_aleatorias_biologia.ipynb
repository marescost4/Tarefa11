# 1. Introdução

A probabilidade é uma ferramenta importante para a Biologia porque muitos fenômenos biológicos apresentam variabilidade e não podem ser previstos com certeza para cada observação individual.

Por exemplo, uma semente pode germinar ou não germinar, um indivíduo pode apresentar ou não determinada característica, e uma armadilha pode capturar diferentes números de insetos em diferentes noites.

A utilização de modelos probabilísticos permite representar matematicamente essa variabilidade e estimar a frequência ou a probabilidade de determinados resultados.

Neste trabalho serão estudados três modelos probabilísticos:

- **Bernoulli**, utilizado para experimentos com dois resultados possíveis;
- **Binomial**, utilizado para contar o número de sucessos em um número fixo de ensaios;
- **Poisson**, utilizado para representar a ocorrência de eventos em determinado intervalo de tempo, espaço, área ou outra unidade de observação.

#2. Variável aleatória

Uma variável aleatória é uma variável que associa valores numéricos aos possíveis resultados de um experimento aleatório. Em outras palavras, ela permite representar numericamente um fenômeno cujo resultado não pode ser conhecido com certeza antes da realização do experimento.

Na Biologia, muitos fenômenos apresentam algum grau de incerteza. Por exemplo, ao selecionar uma semente e colocá-la para germinar, não podemos garantir previamente se ela irá germinar ou não. Podemos, então, representar esse resultado por meio de uma variável aleatória.

###Exemplo biológico

Considere o experimento de selecionar uma semente e observar se ela germina após determinado período.

Podemos definir a variável aleatória (X) da seguinte maneira:

* (X = 1): a semente germinou;
* (X = 0): a semente não germinou.

Nesse caso, o experimento aleatório é a observação da germinação de uma semente, enquanto (X) é a variável aleatória associada ao resultado observado.

Os resultados possíveis são, portanto:

[
X \in {0,1}
]

Esse é um exemplo de variável aleatória discreta, pois os resultados possíveis podem ser contados.

#3. Variáveis aleatórias discretas e contínuas

As variáveis aleatórias podem ser classificadas como discretas ou contínuas:

As variáveis discretas estão relacionadas principalmente a contagens e assumem valores isolados, geralmente números inteiros.

Já as variáveis contínuas podem assumir qualquer valor dentro de determinado intervalo, incluindo valores decimais.


###Exemplos na Biologia



| Tipo | Variável | Justificativa |
|---|---|---|
| Discreta | Número de sementes germinadas | É uma contagem e assume valores inteiros. |
| Discreta | Número de insetos capturados | É uma contagem de indivíduos. |
| Discreta | Número de colônias bacterianas | As colônias são contadas individualmente. |
| Contínua | Massa de uma semente | Pode assumir diversos valores dentro de um intervalo. |
| Contínua | Comprimento de uma folha | Pode apresentar valores decimais. |
| Contínua | Tempo de germinação | Pode ser medido com diferentes valores, inclusive decimais. |


Assim, uma variável como o número de sementes germinadas não pode assumir, por exemplo, 10,5 sementes. Já o comprimento de uma folha pode ser 10,5 cm, 10,51 cm ou outro valor dentro de um intervalo.

Essa diferença é importante porque diferentes modelos probabilísticos podem ser utilizados dependendo do tipo de variável e do fenômeno estudado.

# 4. Modelo de Bernoulli

O modelo de Bernoulli é utilizado quando um experimento apresenta apenas dois resultados possíveis. Esses resultados podem ser chamados de sucesso e fracasso.

Neste trabalho, vamos considerar como exemplo a germinação de sementes.

* Sucesso: a semente germina;
* Fracasso: a semente não germina.

Vamos considerar, para fins de simulação, que a probabilidade de uma semente germinar seja de 80%.

Assim:

[p = 0,80]

e a probabilidade de não germinar é:

[1-p = 0,20]

A variável aleatória (X) pode ser representada por:

[X \sim Bernoulli(0,80)]

com:

[X =
1, se a semente germinar e
0, se a semente não germinar]

É importante destacar que o valor de 0,80 utilizado aqui é simulado, não sendo uma estimativa obtida de uma população real.

```python
# Probabilidade simulada de germinação
p_germinacao = 0.80

# Simulação de uma semente
resultado = bernoulli.rvs(p_germinacao)

print("Resultado do experimento:")

if resultado == 1:
    print("A semente germinou.")
else:
    print("A semente não germinou.")
```

```python
# Simulando 100 sementes
n_sementes = 100

resultados = bernoulli.rvs(
    p_germinacao,
    size=n_sementes
)

numero_germinadas = resultados.sum()
proporcao_germinacao = resultados.mean()

print("Número de sementes:", n_sementes)
print("Sementes germinadas:", numero_germinadas)
print("Proporção observada:", proporcao_germinacao)
```

```python
contagem = pd.Series(resultados).value_counts().sort_index()

plt.bar(
    ["Não germinou", "Germinou"],
    contagem.values
)

plt.xlabel("Resultado")
plt.ylabel("Número de sementes")
plt.title("Resultado da simulação de germinação")
plt.show()
```



A simulação representa 100 sementes submetidas a uma situação em que a probabilidade de germinação foi definida como 80%. Como se trata de uma simulação aleatória, a proporção observada não precisa ser exatamente igual a 80%.

Esse resultado mostra uma característica importante da probabilidade: mesmo conhecendo uma probabilidade teórica, os resultados de uma amostra podem variar devido ao acaso.

Nesse exemplo, cada semente representa um experimento de Bernoulli, pois existem apenas dois resultados possíveis: germinar ou não germinar.

#5. Modelo Binomial

O modelo Binomial pode ser utilizado quando queremos determinar o número de sucessos em um número fixo de ensaios de Bernoulli.

No exemplo das sementes, podemos considerar uma amostra fixa de 20 sementes e observar quantas delas germinam.

Para utilizar o modelo Binomial, temos:

* número fixo de observações;
* dois resultados possíveis em cada observação;
* probabilidade de sucesso definida;
* independência aproximada entre as observações;
* interesse no número total de sucessos.

Vamos considerar:

[
n=20
]

e:

[
p=0,80
]

Assim:

[
X \sim Binomial(20,0,80)
]

A variável (X) representa o número de sementes germinadas entre as 20 sementes analisadas.

Os possíveis valores de (X) são:

[
0,1,2,\ldots,20
]

O valor esperado de uma distribuição Binomial é:

[
E(X)=np
]

Portanto:

[
E(X)=20(0,80)=16
]

Assim, esperamos, em média, 16 sementes germinadas em uma amostra de 20 sementes, considerando as condições definidas na simulação.

```python
n = 20
p = 0.80

prob_15 = binom.pmf(15, n, p)

print(
    f"Probabilidade de exatamente 15 sementes germinarem: "
    f"{prob_15:.4f}"
)
```

```python
prob_pelo_menos_15 = binom.sf(14, n, p)

print(
    f"Probabilidade de pelo menos 15 sementes germinarem: "
    f"{prob_pelo_menos_15:.4f}"
)
```

```python
valor_esperado = n * p

print(
    f"Número esperado de sementes germinadas: "
    f"{valor_esperado:.2f}"
)
```

```python
# Simulando 1000 experimentos,
# cada um com 20 sementes
simulacoes_binomial = binom.rvs(
    n=n,
    p=p,
    size=1000
)

print(
    "Média observada nas simulações:",
    simulacoes_binomial.mean()
)
```

```python
valores = np.arange(0, n + 1)
probabilidades = binom.pmf(valores, n, p)

plt.bar(valores, probabilidades)

plt.xlabel("Número de sementes germinadas")
plt.ylabel("Probabilidade")
plt.title("Distribuição Binomial - Germinação de sementes")
plt.show()
```


Nesse exemplo, o modelo Binomial permite analisar o número de sementes que germinam em uma amostra fixa de 20 sementes.

O valor esperado é de 16 sementes germinadas. Isso não significa que toda amostra terá exatamente 16 sementes germinadas, mas que esse é o número médio esperado quando o experimento é repetido muitas vezes sob as mesmas condições.

A utilização do modelo depende de algumas suposições. É necessário considerar um número fixo de sementes, dois resultados possíveis para cada semente e uma probabilidade de germinação aproximadamente constante. Além disso, a independência entre as sementes deve ser uma aproximação razoável.

Na prática, essa independência pode ser afetada por fatores ambientais compartilhados, como temperatura, umidade e qualidade do substrato. Portanto, o modelo é uma representação simplificada do processo biológico.

#6. Modelo de Poisson

O modelo de Poisson é utilizado para representar o número de vezes que determinado evento ocorre em um intervalo definido de tempo, espaço, área, volume ou outra unidade de observação.

Neste exemplo, vamos considerar o número de insetos capturados por uma armadilha durante uma noite.

Suponha que, em condições semelhantes de monitoramento, sejam capturados em média 5 insetos por noite.

Essa média é representada pelo parâmetro:

[
\lambda=5
]

Assim:

[
X \sim Poisson(5)
]

A variável aleatória (X) representa o número de insetos capturados por uma armadilha durante uma noite.

Os possíveis valores são:

[
0,1,2,3,\ldots
]

Diferentemente do modelo Binomial, não existe um número máximo previamente estabelecido de insetos que podem ser capturados.

Neste exemplo, a unidade de observação é uma armadilha durante uma noite, e a taxa média de ocorrência é de 5 insetos por armadilha/noite.

Os dados utilizados são simulados para representar um cenário de monitoramento ambiental.

```python
lambda_insetos = 5

prob_3 = poisson.pmf(3, lambda_insetos)

print(
    f"Probabilidade de capturar exatamente 3 insetos: "
    f"{prob_3:.4f}"
)
```

```python
prob_zero = poisson.pmf(0, lambda_insetos)

print(
    f"Probabilidade de não capturar nenhum inseto: "
    f"{prob_zero:.4f}"
)
```

```python
n_noites = 100

capturas = poisson.rvs(
    mu=lambda_insetos,
    size=n_noites
)

dados_insetos = pd.DataFrame({
    "Noite": np.arange(1, n_noites + 1),
    "Insetos_capturados": capturas
})

dados_insetos.head()
```

```python
media_observada = dados_insetos["Insetos_capturados"].mean()

print(
    f"Média de insetos capturados nas 100 noites: "
    f"{media_observada:.2f}"
)
```

```python
valores_poisson = np.arange(0, capturas.max() + 1)

probabilidades_poisson = poisson.pmf(
    valores_poisson,
    lambda_insetos
)

plt.bar(
    valores_poisson,
    probabilidades_poisson
)

plt.xlabel("Número de insetos capturados por noite")
plt.ylabel("Probabilidade")
plt.title("Distribuição de Poisson - Captura de insetos")
plt.show()
```



Neste exemplo, o modelo de Poisson foi utilizado para representar o número de insetos capturados por uma armadilha durante uma noite.

O parâmetro (\lambda=5) representa uma média de cinco insetos capturados por armadilha por noite. A simulação de 100 noites apresenta pequenas variações em relação a essa média, o que é esperado devido à aleatoriedade do processo.

Esse tipo de modelo pode ser útil em estudos de monitoramento ambiental, nos quais pesquisadores registram quantos indivíduos são observados em determinado período ou unidade de amostragem.

Entretanto, a aplicação do modelo depende das características dos dados. Se houver forte agregação dos insetos, mudanças importantes na taxa de captura entre noites ou dependência entre os eventos, o modelo de Poisson pode não representar adequadamente o processo biológico.

#7. Comparação entre os modelos

Os modelos de Bernoulli, Binomial e Poisson podem ser utilizados para representar diferentes situações probabilísticas encontradas na Biologia.

Modelo	Fenômeno representado	Valores possíveis	Parâmetros	Exemplo biológico
Bernoulli	Um único experimento com dois resultados	0 ou 1	(p)	Uma semente germina ou não
Binomial	Número de sucessos em um número fixo de ensaios	0 até (n)	(n,p)	Número de sementes germinadas em 20 sementes
Poisson	Número de eventos em um intervalo definido	0, 1, 2, 3…	(\lambda)	Insetos capturados por armadilha durante uma noite

O modelo de Bernoulli pode ser considerado o caso mais simples dos três, pois envolve apenas uma observação com dois resultados possíveis.

O modelo Binomial amplia essa situação para vários ensaios, mantendo um número fixo de observações e dois resultados possíveis em cada uma delas.

Já o modelo de Poisson é utilizado para contagens de eventos em determinada unidade de tempo, espaço, área, volume ou outra unidade de observação. Por isso, não exige que exista previamente um número máximo de eventos.

Assim, a escolha do modelo deve considerar o tipo de fenômeno biológico que está sendo estudado e as características dos dados.

```python
# Bernoulli
bernoulli_sim = bernoulli.rvs(
    p=0.8,
    size=1000
)

# Binomial
binomial_sim = binom.rvs(
    n=20,
    p=0.8,
    size=1000
)

# Poisson
poisson_sim = poisson.rvs(
    mu=5,
    size=1000
)

print("Bernoulli - média:", bernoulli_sim.mean())
print("Binomial - média:", binomial_sim.mean())
print("Poisson - média:", poisson_sim.mean())
```



As simulações mostram que os três modelos possuem comportamentos diferentes porque representam situações diferentes.

No modelo de Bernoulli, cada observação pode resultar apenas em 0 ou 1. No modelo Binomial, os valores representam o número de sucessos em um conjunto fixo de 20 ensaios. Já no modelo de Poisson, os valores representam a quantidade de eventos observados em uma unidade definida, neste caso, uma noite de captura.

As médias observadas nas simulações podem apresentar pequenas diferenças em relação aos valores teóricos. Isso acontece porque os resultados das simulações são influenciados pelo acaso. Com um número maior de repetições, a média observada tende a se aproximar do valor esperado pelo modelo.

#8. Limitações das aplicações

Os exemplos apresentados neste notebook foram construídos por meio de simulações e não representam necessariamente uma população biológica específica.

No caso da germinação, foi utilizada uma probabilidade fixa de 80%. Na realidade, a germinação pode ser influenciada por diversos fatores, como temperatura, disponibilidade de água, qualidade das sementes, luminosidade e características do substrato.

Da mesma forma, no exemplo dos insetos, foi considerada uma taxa média constante de cinco indivíduos por armadilha por noite. Na prática, a quantidade de insetos capturados pode variar de acordo com temperatura, precipitação, horário, sazonalidade, disponibilidade de alimento, localização da armadilha e características do ambiente.

Além disso, a independência entre observações pode não ser completamente válida em situações biológicas reais. Por isso, os modelos devem ser entendidos como representações matemáticas de determinados fenômenos e sua adequação deve ser avaliada de acordo com os dados e com as características do sistema estudado.

#9. Conclusão

A utilização da Probabilidade na Biologia permite representar e estudar situações que apresentam variabilidade e incerteza. Neste trabalho, foi possível observar como o conceito de variável aleatória pode ser aplicado a diferentes fenômenos biológicos e como as variáveis podem ser classificadas em discretas ou contínuas.

Os modelos de Bernoulli, Binomial e Poisson apresentam aplicações diferentes. O modelo de Bernoulli é adequado para situações com apenas dois resultados possíveis, como uma semente germinar ou não. O modelo Binomial permite analisar o número de sucessos em um conjunto fixo de observações, como o número de sementes germinadas entre 20 sementes. Já o modelo de Poisson permite estudar a ocorrência de eventos em uma determinada unidade, como o número de insetos capturados por uma armadilha durante uma noite.

As simulações também mostraram que os resultados observados podem variar devido ao acaso, mesmo quando os parâmetros do modelo são conhecidos. Por isso, a probabilidade não determina exatamente o resultado de um experimento individual, mas permite compreender e quantificar padrões esperados quando os experimentos são realizados repetidamente.

Por fim, é importante destacar que os modelos probabilísticos são simplificações da realidade. Em estudos biológicos reais, fatores ambientais e características dos organismos podem interferir nos resultados e fazer com que algumas das suposições dos modelos não sejam completamente atendidas. Dessa forma, além de realizar os cálculos, é necessário avaliar se o modelo escolhido representa de maneira razoável o fenômeno biológico estudado.
